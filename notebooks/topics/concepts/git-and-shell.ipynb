Use a temporary repository to inspect an input change without touching the handbook checkout.

## Setup and sample data

Create a disposable Git repository and a CSV file. The cell uses the Git executable available in Colab.

```python
from pathlib import Path
import subprocess
from tempfile import TemporaryDirectory

workspace = TemporaryDirectory()
root_dir = Path(workspace.name)


def git(*args):
    return subprocess.run(["git", *args], cwd=root_dir, check=True, capture_output=True, text=True).stdout


git("init", "--quiet")
(root_dir / "orders.csv").write_text("id,cents\n1,1200\n")
print(git("status", "--short"))
```

## Inspect exactly what will be committed

git diff shows unstaged changes to tracked files. git diff --cached shows the staged content. Shell redirection can overwrite files, so first confirm the working directory and inspect the diff.

**Where it matters**

Review an input schema or job configuration before opening a PR.

**Watch out for**

A clean unstaged diff can still leave staged changes. Check both views.

```python
git("add", "--", "orders.csv")
print(git("diff", "--cached", "--", "orders.csv"))
```

**What to notice:** The staged diff contains the CSV header and one data row.
