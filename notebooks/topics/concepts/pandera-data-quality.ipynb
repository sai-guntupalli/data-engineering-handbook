## Setup and sample data

The fixture has a duplicate order ID and a negative amount.

```python
import pandas as pd
import pandera.pandas as pa

incoming = pd.DataFrame({"order_id": [1, 1, 2], "cents": [100, 200, -1]})
```

## Validate before publishing

A schema describes required columns, types and value rules. Lazy validation collects several failures in one pass.

**Where it matters**

Return actionable failures before promoting a candidate to a published table.

**Watch out for**

A contract checks the rows it sees. Reconcile completeness against independent input counts or a manifest.

```python
schema = pa.DataFrameSchema({
    "order_id": pa.Column(int, unique=True),
    "cents": pa.Column(int, checks=pa.Check.ge(0)),
}, strict=True)

try:
    schema.validate(incoming, lazy=True)
except pa.errors.SchemaErrors as error:
    print(error.failure_cases[["column", "failure_case"]].to_string(index=False))
```

**What to notice:** The duplicate key 1 and negative cents value -1 appear in the failure report.
