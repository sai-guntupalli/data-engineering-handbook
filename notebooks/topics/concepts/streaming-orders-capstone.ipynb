An in-memory model connects replay, keyed writes and a correct streaming total.

## Setup and sample data

An at-least-once consumer receives one repeated event after restarting. Each event has a stable ID and a paid amount.

```python
arrivals = [(1, 1200), (1, 1200), (2, 800)]
```

## Replay from a durable boundary

The checkpoint may lag behind the sink, causing an event to arrive twice. This model stores the result by event ID and computes revenue from the keyed records.

**Where it matters**

Make an external sink replay-safe when processing and checkpointing are separate.

**Watch out for**

The dictionary represents a durable keyed table. In production the sink must persist its keys across worker failures.

```python
sink = {}
for event_id, cents in arrivals:
    sink.setdefault(event_id, cents)
    print("after event", event_id, "revenue:", sum(sink.values()))
```

**What to notice:** Revenue stays at 1200 cents on replay and reaches 2000 cents after event 2.
