The earlier chapters are general Python. This one is about the habits that make a data pipeline correct when it runs
again, when it is partly wrong and when nobody is watching:

- **Idempotent**: running a job twice gives the same result as running it once.
- **Restartable**: a crash leaves no half-finished state, and a rerun picks up where the last good run ended.
- **Validated**: bad records are counted and kept aside, not silently dropped or allowed to break the load.
- **Observable**: every run logs what it read, loaded and rejected.

Each idea is built here with the standard library only, so every cell runs on Colab and locally. For columnar files
see [Parquet](parquet.ipynb), for scheduling see [Airflow](airflow.ipynb), and for quality rules see

## Configuration from the environment

Pipelines run in many places, so settings come from environment variables with safe defaults. Validate them once, at
startup, and fail with a clear message. A typo such as `BATCH_SIZE=0` should stop the job before it reads a single row.

### Where you'll use it

- **Twelve-factor deployments:** the same image runs in dev, staging and production with different environment variables.
- **Kubernetes and Airflow:** settings arrive as environment variables or secrets, not hard-coded values.

```python
import os

def load_config(env):
    config = {
        "batch_size": int(env.get("BATCH_SIZE", "500")),
        "max_retries": int(env.get("MAX_RETRIES", "3")),
        "landing_dir": env.get("LANDING_DIR", "/data/landing"),
    }
    if config["batch_size"] < 1:
        raise ValueError("BATCH_SIZE must be at least 1")
    if config["max_retries"] < 1:
        raise ValueError("MAX_RETRIES must be at least 1")
    return config

assert load_config({})["batch_size"] == 500
assert load_config({"BATCH_SIZE": "50"})["batch_size"] == 50

try:
    load_config({"BATCH_SIZE": "0"})
except ValueError as error:
    assert "BATCH_SIZE" in str(error)
else:
    raise AssertionError("expected ValueError")
```

## Validate records and keep the rejects

A record is either loaded or **quarantined** with the reason. Quarantine keeps the bad data for investigation and keeps
counts honest: `read = loaded + quarantined + skipped` should always hold. Money is parsed from text with `Decimal`
and stored as integer cents, so there is no float error.

### Where you'll use it

- **Data contracts:** a validation layer that rejects rows violating the agreed schema and keeps them for the data producer to fix.
- **Compliance:** keeping rejected rows in a quarantine table with a reason, so every row can be traced.

```python
import json
from decimal import Decimal, InvalidOperation

def parse_amount_cents(text):
    try:
        amount = Decimal(str(text))
    except InvalidOperation as error:        # Decimal raises InvalidOperation, not ValueError
        raise ValueError(f"not a number: {text!r}") from error
    return int((amount * 100).to_integral_value())

def parse_record(raw):
    order_id = int(raw["order_id"])
    amount_cents = parse_amount_cents(raw["amount"])
    updated_at = str(raw["updated_at"])
    if amount_cents < 0:
        raise ValueError("negative amount")
    if not updated_at:
        raise ValueError("missing updated_at")
    return {"order_id": order_id, "amount_cents": amount_cents, "updated_at": updated_at}

good = parse_record(json.loads('{"order_id": "7", "amount": "12.50", "updated_at": "2024-01-15T10:00:00"}'))
assert good == {"order_id": 7, "amount_cents": 1250, "updated_at": "2024-01-15T10:00:00"}

for bad_line in ['{"order_id": "x", "amount": "1", "updated_at": "t"}',
                 '{"order_id": 1, "amount": "oops", "updated_at": "t"}',
                 '{"order_id": 1, "amount": "-3", "updated_at": "t"}',
                 '{"order_id": 1, "amount": "2"}']:
    try:
        parse_record(json.loads(bad_line))
    except (KeyError, ValueError):
        pass
    else:
        raise AssertionError(f"should have been rejected: {bad_line}")
```

## Idempotent loads with an upsert

An **upsert** inserts a row, or updates it when the key already exists. Loading the same batch twice then leaves the
table unchanged. The guard in the `WHERE` clause also stops an **older** version of a record from overwriting a newer
one, which happens when late data arrives out of order.

This uses SQLite, which ships with Python. The same `INSERT ... ON CONFLICT` pattern exists in PostgreSQL.

### Where you'll use it

- **Retrying failed loads:** an Airflow task that retries after a timeout must not double-count orders.
- **Change data capture targets:** a sink receiving inserts and updates from a CDC stream applies them with upserts, using the same pattern.

```python
import sqlite3

def make_db():
    db = sqlite3.connect(":memory:")
    db.execute("""
        CREATE TABLE orders (
            order_id     INTEGER PRIMARY KEY,
            amount_cents INTEGER NOT NULL,
            updated_at   TEXT NOT NULL
        )
    """)
    return db

def upsert(db, records):
    db.executemany(
        """
        INSERT INTO orders (order_id, amount_cents, updated_at) VALUES (:order_id, :amount_cents, :updated_at)
        ON CONFLICT(order_id) DO UPDATE SET
            amount_cents = excluded.amount_cents,
            updated_at   = excluded.updated_at
        WHERE excluded.updated_at >= orders.updated_at
        """,
        records,
    )
    db.commit()

db = make_db()
batch = [
    {"order_id": 1, "amount_cents": 500, "updated_at": "2024-01-10"},
    {"order_id": 2, "amount_cents": 900, "updated_at": "2024-01-11"},
]
upsert(db, batch)
upsert(db, batch)                                   # running it again changes nothing
assert db.execute("SELECT COUNT(*) FROM orders").fetchone()[0] == 2

upsert(db, [{"order_id": 1, "amount_cents": 700, "updated_at": "2024-01-12"}])   # newer: applied
upsert(db, [{"order_id": 1, "amount_cents": 1, "updated_at": "2024-01-01"}])     # older: ignored
assert db.execute("SELECT amount_cents FROM orders WHERE order_id = 1").fetchone()[0] == 700
```

## Incremental processing with a watermark

Reprocessing everything every night is expensive. A **watermark** is the newest `updated_at` already loaded. Each run
reads only records newer than the watermark, and moves the watermark forward **only after the load succeeds**. If the
run crashes, the watermark stays put and the next run retries the same records.

### Where you'll use it

- **Nightly loads:** load only rows changed since the last successful run, instead of the full table.
- **Streaming equivalents:** Spark Structured Streaming and Flink track the same concept as a watermark for late events.

```python
def run_pipeline(raw_lines, db, state, run_id):
    watermark = state.get("watermark", "")
    loaded, skipped, quarantined = [], 0, []

    for line in raw_lines:
        try:
            record = parse_record(json.loads(line))
        except (json.JSONDecodeError, KeyError, ValueError) as error:
            quarantined.append({"line": line, "reason": str(error)})
            continue
        if record["updated_at"] <= watermark:
            skipped += 1
            continue
        loaded.append(record)

    if loaded:
        upsert(db, loaded)
        state["watermark"] = max(r["updated_at"] for r in loaded)   # only after the commit

    summary = {
        "run_id": run_id,
        "read": len(raw_lines),
        "loaded": len(loaded),
        "skipped": skipped,
        "quarantined": len(quarantined),
    }
    assert summary["read"] == summary["loaded"] + summary["skipped"] + summary["quarantined"]
    return summary, quarantined

lines = [
    '{"order_id": 1, "amount": "5.00", "updated_at": "2024-01-10"}',
    '{"order_id": 2, "amount": "oops", "updated_at": "2024-01-11"}',
    '{"order_id": 3, "amount": "8.25", "updated_at": "2024-01-12"}',
]
db = make_db()
state = {}
summary, rejects = run_pipeline(lines, db, state, run_id="run-1")
assert summary == {"run_id": "run-1", "read": 3, "loaded": 2, "skipped": 0, "quarantined": 1}
assert rejects[0]["line"] == lines[1]
assert state["watermark"] == "2024-01-12"

# Rerun the same file: everything is at or below the watermark, so nothing loads twice.
summary, _ = run_pipeline(lines, db, state, run_id="run-2")
assert summary["loaded"] == 0 and summary["skipped"] == 2
assert db.execute("SELECT COUNT(*) FROM orders").fetchone()[0] == 2
```

## Retries with exponential backoff

Network and database calls fail transiently. Retry them, but only the errors that can succeed later, and wait longer
after each failure. Adding random **jitter** stops many workers from retrying at the same moment.

The decorator takes its `sleep` and random source as arguments. Tests pass fakes, so the delays are checked exactly
and the test does not actually wait.

### Where you'll use it

- **Calling flaky services:** a database restart or a 503 from a SaaS API is retried with growing waits.
- **Cloud SDKs:** AWS and GCP client libraries implement this pattern internally, and their retry settings use the same ideas.

```python
import functools

def retry(attempts=3, base_delay=0.5, retry_on=(ConnectionError,), sleep=None, jitter=None):
    sleep = sleep or __import__("time").sleep
    jitter = jitter or __import__("random").random

    def decorator(fn):
        @functools.wraps(fn)
        def wrapper(*args, **kwargs):
            for attempt in range(1, attempts + 1):
                try:
                    return fn(*args, **kwargs)
                except retry_on:
                    if attempt == attempts:
                        raise
                    # exponential backoff, scaled by a jitter factor between 0.5 and 1.5
                    sleep(base_delay * 2 ** (attempt - 1) * (0.5 + jitter()))
        return wrapper
    return decorator

waits = []
calls = {"n": 0}

@retry(attempts=4, base_delay=1.0, sleep=waits.append, jitter=lambda: 0.5)   # jitter factor 1.0
def flaky_insert():
    calls["n"] += 1
    if calls["n"] < 3:
        raise ConnectionError("database restarting")
    return "inserted"

assert flaky_insert() == "inserted"
assert waits == [1.0, 2.0]          # 1 s, then 2 s: doubling each time

@retry(attempts=2, sleep=waits.append, jitter=lambda: 0.5)
def bad_request():
    raise ValueError("malformed input")          # not retryable: raised at once

try:
    bad_request()
except ValueError:
    assert len(waits) == 2                        # no extra waits were added
```

## Writing partitioned output safely

Analytics engines read folders split by a key, such as `dt=2024-01-15/`. This layout lets a query skip whole folders.
Write each partition to a temporary file and move it into place, so a crash never leaves a partial file that a reader
could mistake for a complete one.

### Where you'll use it

- **Data lakes:** Hive-style `dt=` folders are how Spark, Athena, BigQuery external tables and Trino prune data.
- **Backfills:** rewriting one partition without touching the others.

```python
import tempfile
from pathlib import Path

def write_partition(root, day, records):
    folder = Path(root) / f"dt={day}"
    folder.mkdir(parents=True, exist_ok=True)
    target = folder / "part-0000.jsonl"
    tmp = folder / "part-0000.jsonl.tmp"
    with tmp.open("w", encoding="utf-8") as handle:
        for record in records:
            handle.write(json.dumps(record, sort_keys=True) + "\n")
    tmp.replace(target)                       # atomic on one filesystem
    return target

warehouse = Path(tempfile.mkdtemp(prefix="warehouse-"))
write_partition(warehouse, "2024-01-15", [{"order_id": 1}])
write_partition(warehouse, "2024-01-16", [{"order_id": 2}, {"order_id": 3}])

partitions = sorted(p.parent.name for p in warehouse.rglob("*.jsonl"))
assert partitions == ["dt=2024-01-15", "dt=2024-01-16"]
assert not list(warehouse.rglob("*.tmp"))
```

## Detecting changes with a content hash

To find which rows changed since the last load, hash a canonical form of each record. Canonical means the same
key order and the same formatting every time, so equal content gives an equal hash.

### Where you'll use it

- **Snapshot diffs:** comparing today's table to yesterday's by hash to find updated rows without comparing every column.
- **Deduplication across sources:** hashing the business columns to detect the same record arriving from two systems.

```python
import hashlib

def row_hash(record):
    canonical = json.dumps(record, sort_keys=True, separators=(",", ":"))
    return hashlib.sha256(canonical.encode("utf-8")).hexdigest()

before = {"order_id": 1, "amount_cents": 500}
after = {"amount_cents": 500, "order_id": 1}          # same content, different key order
changed = {"order_id": 1, "amount_cents": 650}

assert row_hash(before) == row_hash(after)
assert row_hash(before) != row_hash(changed)
assert len(row_hash(before)) == 64
```

## Deduplicating: keep the latest version

Sources send the same record more than once: retries, replays and corrections. Deduplicate on the **business key**,
keeping the version with the newest `updated_at`. Ties keep the first one seen, so the result is deterministic.

### Where you'll use it

- **At-least-once sources:** Kafka and most message queues redeliver messages, so every consumer needs this step.
- **Late corrections:** a source resends a corrected order, and the pipeline keeps only the newest version.

```python
def latest_per_key(records, key, timestamp):
    latest = {}
    for record in records:
        current = latest.get(record[key])
        if current is None or record[timestamp] > current[timestamp]:
            latest[record[key]] = record
    return list(latest.values())

events = [
    {"order_id": 1, "status": "placed", "updated_at": "2024-01-10"},
    {"order_id": 1, "status": "shipped", "updated_at": "2024-01-12"},
    {"order_id": 2, "status": "placed", "updated_at": "2024-01-11"},
    {"order_id": 1, "status": "placed", "updated_at": "2024-01-10"},   # a replay of the first event
]
result = sorted(latest_per_key(events, "order_id", "updated_at"), key=lambda r: r["order_id"])
assert result == [
    {"order_id": 1, "status": "shipped", "updated_at": "2024-01-12"},
    {"order_id": 2, "status": "placed", "updated_at": "2024-01-11"},
]
```

## Batches with bounded memory and a run summary

Process a large input in batches and log progress per batch. Log the summary at the end of the run, with the run id, so
an operator can find it and compare runs.

### Where you'll use it

- **Large backfills:** processing a year of history in fixed-size batches so the worker's memory never grows with the input.
- **Run reports:** the summary line is what a monitoring dashboard or a Slack alert reads after each run.

```python
import logging
import io

def chunked(iterable, size):
    batch = []
    for item in iterable:
        batch.append(item)
        if len(batch) == size:
            yield batch
            batch = []
    if batch:
        yield batch

stream = io.StringIO()
logger = logging.getLogger("pipeline.orders")
logger.handlers.clear()
logger.addHandler(logging.StreamHandler(stream))
logger.setLevel(logging.INFO)

def process_in_batches(records, size, run_id):
    total = 0
    for number, batch in enumerate(chunked(records, size), start=1):
        total += len(batch)
        logger.info("run=%s batch=%d rows=%d", run_id, number, len(batch))
    logger.info("run=%s done rows=%d", run_id, total)
    return total

assert process_in_batches(range(5), 2, "run-7") == 5
logged = stream.getvalue().splitlines()
assert logged[0] == "run=run-7 batch=1 rows=2"
assert logged[-1] == "run=run-7 done rows=5"
```

## Practice questions

1. Show that a record with a missing `amount` is quarantined rather than crashing the run, and that the counts still
   add up.
2. Explain why the watermark is moved **after** `upsert` rather than before it.
3. Write `file_name_for(source, day, sequence)` that returns `"orders_2024-01-15_0003.jsonl"` for `("orders", "2024-01-15", 3)`.

## Answers

### Question 1

`parse_record` already raises `KeyError` for a missing field, and `run_pipeline` catches `KeyError` with the other
parse errors. So the record is quarantined, and the three counts still add up.

```python
db = make_db()
state = {}
summary, rejects = run_pipeline(
    ['{"order_id": 9, "updated_at": "2024-01-20"}',
     '{"order_id": 10, "amount": "1.00", "updated_at": "2024-01-20"}'],
    db, state, run_id="run-3")
assert summary["quarantined"] == 1 and summary["loaded"] == 1
assert summary["read"] == summary["loaded"] + summary["skipped"] + summary["quarantined"]
```

### Question 2

If the upsert fails after the watermark moved, the next run skips the records that were never written. Moving the
watermark last means a failure leads to a retry, and the idempotent upsert makes the retry safe.

### Question 3

```python
def file_name_for(source, day, sequence):
    return f"{source}_{day}_{sequence:04d}.jsonl"

assert file_name_for("orders", "2024-01-15", 3) == "orders_2024-01-15_0003.jsonl"
```

## Interview notes

- **Idempotency is the core property.** Design every load so a rerun produces the same state, using upserts keyed by a
  business key, or overwriting whole partitions.
- **Move state last.** Advance the watermark or the offset only after the data is committed.
- **Quarantine, do not drop.** Rejected rows with a reason let you measure data quality and fix the source.
- **Retry only transient errors**, with backoff and jitter. Retrying a bad request wastes time and can flood a service.
- **Store money as integers**, and make counts reconcile: `read = loaded + skipped + quarantined`.
