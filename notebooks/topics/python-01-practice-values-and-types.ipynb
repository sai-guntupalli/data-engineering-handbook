from decimal import Decimal

def sum_money(amounts):
    total = sum((Decimal(a.replace(",", "")) for a in amounts), Decimal("0"))
    return str(total.quantize(Decimal("0.01")))

def mask_middle(text):
    if len(text) < 3:
        return text
    return text[0] + "*" * (len(text) - 2) + text[-1]
