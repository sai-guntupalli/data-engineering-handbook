Dynamic programming solves a problem by reusing answers to smaller versions of it. The skill is writing the recurrence first, then choosing the order to fill it in. Statements are paraphrased, and each solution is collapsed under its exercise.

<details>
<summary><b>How to use this notebook</b></summary>

Work through the exercises in order. Each one has three parts:

1. **Question:** what to build, with the rules it must follow.
2. **Exercise cell:** replace the `pass` with your code. Keep the function name and parameters.
3. **Check cell:** run it after your code. Each line says `pass` when your function gives the expected result, and `todo` when it does not yet. Checks never stop the notebook, so you can run everything at any time.

The last cell prints your score. Use `Runtime -> Run all` to rerun everything after a change.

</details>

````python
_tally = {"passed": 0, "total": 0}

def check(label, make_actual, expected):
    """Run one check. It never raises, so the notebook always runs from top to bottom."""
    _tally["total"] += 1
    try:
        actual = make_actual()
    except Exception as error:
        print(f"error  {label}: {type(error).__name__}: {error}")
        return
    if actual == expected:
        _tally["passed"] += 1
        print(f"pass   {label}")
    else:
        print(f"todo   {label}: got {actual!r}")

def raises(make_actual, error_type):
    """True when the call raises error_type, False when it raises anything else or nothing."""
    try:
        make_actual()
    except error_type:
        return True
    except Exception:
        return False
    return False

def score():
    print(f"\nScore: {_tally['passed']} of {_tally['total']} checks pass")
````

```python
def climb_stairs(n):
    # Approach: the number of ways to reach step k is ways(k-1) + ways(k-2), because the last
    # move is one step or two. That is Fibonacci. Keep only the last two values.
    # Complexity: O(n) time, O(1) space.
    before, current = 1, 1                    # ways to reach step 0 and step 1
    for _ in range(n - 1):
        before, current = current, before + current
    return current
```

```python
def rob(nums):
    # Approach: at each house, either rob it (taken = skipped_previous + value) or skip it.
    # Track the best total when the previous house was robbed (taken) and when it was not
    # (skipped). Two variables are enough because each house depends only on the one before.
    # Complexity: O(n) time, O(1) space.
    taken, skipped = 0, 0
    for value in nums:
        taken, skipped = skipped + value, max(taken, skipped)   # rob this, or keep the best so far
    return max(taken, skipped)
```

```python
def coin_change(coins, amount):
    # Approach: bottom-up DP. best[t] = fewest coins that make total t.
    # For each total, try every coin that fits and build on best[total - coin].
    # amount + 1 is a safe "impossible" value: no real answer can need more coins than that.
    # Complexity: O(amount * len(coins)) time, O(amount) space.
    INFEASIBLE = amount + 1
    best = [0] + [INFEASIBLE] * amount       # best[0] = 0 coins for total 0
    for total in range(1, amount + 1):
        for coin in coins:
            if coin <= total:
                best[total] = min(best[total], best[total - coin] + 1)
    return best[amount] if best[amount] != INFEASIBLE else -1
```

```python
def length_of_lis(nums):
    # Approach: lengths[i] = longest increasing subsequence that ends at nums[i].
    # Look back at every earlier j with a smaller value and extend its best run by one.
    # The answer is the largest value in lengths.
    # Complexity: O(n^2) time, O(n) space. (A patience-sorting method gives O(n log n).)
    lengths = [1] * len(nums)                # every element alone is a subsequence of length 1
    for i in range(len(nums)):
        for j in range(i):
            if nums[j] < nums[i]:            # strictly increasing: extend j's run
                lengths[i] = max(lengths[i], lengths[j] + 1)
    return max(lengths, default=0)           # default handles an empty input
```

```python
def longest_common_subsequence(first, second):
    # Approach: a 2D table where table[i][j] = LCS length of first[:i] and second[:j].
    # If the characters match, extend the diagonal (table[i-1][j-1] + 1).
    # Otherwise drop one character from either string and keep the better result.
    # Complexity: O(m * n) time and space.
    table = [[0] * (len(second) + 1) for _ in range(len(first) + 1)]   # row/col 0 = empty prefix
    for i, a in enumerate(first, 1):
        for j, b in enumerate(second, 1):
            if a == b:
                table[i][j] = table[i - 1][j - 1] + 1
            else:
                table[i][j] = max(table[i - 1][j], table[i][j - 1])
    return table[-1][-1]
```
