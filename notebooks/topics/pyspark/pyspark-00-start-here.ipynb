*Author: [Sai Guntupalli](https://github.com/sai-guntupalli) · [sai-guntupalli/data-engineering-handbook](https://github.com/sai-guntupalli/data-engineering-handbook) · CC BY 4.0*

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/sai-guntupalli/data-engineering-handbook/blob/master/notebooks/topics/pyspark/pyspark-00-start-here.ipynb)

This course teaches PySpark the way data engineers use it: read raw files, clean and join them, publish tables
that stay correct when a job is retried, process data as it arrives, and make slow jobs fast. It is written for two
readers: someone learning Spark for the first time, and someone preparing for a data-engineering interview.

Every lesson runs on your laptop or in Google Colab with no cluster. The examples use the same small tables
throughout, so you can compare results from one lesson to the next.

*Last reviewed: November 2026, against Apache Spark 4.0.*

## How to run the notebooks

- **Colab:** open any notebook with the *Open in Colab* badge and choose *Runtime → Run all*. Colab has Java
  already; the first cell installs PySpark.
- **Locally:** you need Python 3.9 or newer and Java 17 or 21. Each notebook installs its own packages in its first
  cell. On macOS, `brew install openjdk@17` and point `JAVA_HOME` at it.
- Run the cells of a notebook **in order**. Each lesson starts its own Spark session and creates its own sample
  data, so lessons do not depend on each other.

## How a lesson is laid out

1. A short scenario: the problem the lesson solves.
2. Sections that each explain **how it works**, **where it matters** and what to **watch out for**.
3. Small runnable examples. Read the output, then the **What to notice** note under it.
4. **Check yourself** questions with answers, and **Key takeaways**.
5. For most lessons, a drills notebook with exercises, expected outputs and hidden reference answers.

## The sample data

A slice of an online shop, created in every lesson with `spark.createDataFrame`:

| table | one row per | what makes it interesting |
| --- | --- | --- |
| `customers` | customer | customer 4 has no region; customers 7 and 8 have never ordered |
| `orders` | order | statuses `placed`, `paid`, `shipped`, `returned`; order 12 lands just after midnight UTC |
| `order_items` | order line | each order's lines add up to its amount; order 9 is a 20-unit bulk order |
| `products` | SKU | `SKU-4014` is missing, so one order line has no product |
| `events` | user event | one event was committed twice by a retry |

Money is stored as `decimal`, never as a floating-point number, so totals add up exactly.
