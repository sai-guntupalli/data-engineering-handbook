## How the notebooks work

- **Run anywhere.** Every notebook opens in Colab and runs top to bottom; locally, install the `notebooks` extra and
  Java 17 or 21. Notebooks 10 and 15 start a Delta Lake session, whose jars download on first use.
- **Shared sample data.** All notebooks use the same small `customers`, `orders`, `order_items`, `products` and
  `events` tables, so you only learn the data once.
- **Practice format.** Each exercise names the reference entries it practises, shows its input and the exact expected
  output, and hides a solution you can open when you are done or stuck. CI checks every solution against its
  published output.
