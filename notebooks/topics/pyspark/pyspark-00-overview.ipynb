[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/sai-guntupalli/data-engineering-handbook/blob/master/notebooks/topics/pyspark/pyspark-00-overview.ipynb)

Build a small orders pipeline and learn to operate it. Start by tracing how Spark executes a query. Then read
and validate files, join them without changing the intended grain, publish a repeatable batch, handle new
events through a stream, and tune the result.

Every module follows the same shape: a scenario, then sections that explain how a feature works, where it
matters and what to watch out for, each ending with a runnable example whose output is explained in a
**What to notice** note. Each module closes with **Check yourself** questions and **Key takeaways**. All modules run on the same four small
tables, and each one runs on its own.

**Local setup:** Python, PySpark 4.0, and Java 17 or 21 (Colab already has Java). Each notebook installs its
own packages. Basic [Python](../python/learning/python-01-values-and-types.ipynb) and
[SQL](../sql/sql-01-basics.ipynb) are enough to start.

*Last reviewed: November 2026, against Apache Spark 4.0.*

## Follow an order through the course

Keep one question in mind as you move from module to module: how do you turn a delivery of raw order events
into a daily revenue table you can trust?

| Step | Example | What to verify | Module |
| --- | --- | --- | --- |
| Read and clean | Parse the order timestamp and amount with a declared schema | Rejected rows are counted next to accepted rows | 2 |
| Join and aggregate | Attach the customer region and sum amounts by day | The join keeps one row per order | 3 |
| Handle updates | Keep the latest event per order using a source sequence | Tied timestamps give a reproducible winner | 4 |
| Publish and operate | Replace only the affected days and rerun the same delivery | Counts and totals match, and a rerun changes nothing | 5, 7 |
