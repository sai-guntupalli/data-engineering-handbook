PySpark is the Python API for Apache Spark. It runs distributed DataFrame and SQL work on a cluster, or locally for learning. The material is split into eight modules, each one a notebook you can run on its own. Work through them in order: each one builds on the last.

Before you start: [Python 1 to 4](../python/learning/python-01-values-and-types.ipynb) and [SQL 1 to 8](../sql/sql-01-basics.ipynb). The notebooks pin PySpark 4.0, which needs Java 17 or later when you run them on your own machine.

## Modules

| # | Module | What it covers |
| --- | --- | --- |
| 1 | [Foundations](pyspark-01-foundations.ipynb) | What PySpark is, sessions, DataFrames versus RDDs, transformations, actions, lazy evaluation |
| 2 | [Reading, SQL and cleaning](pyspark-02-reading-sql-and-cleaning.ipynb) | Explicit schemas, temp views and Spark SQL, cleaning and typing columns |
| 3 | [Joins and aggregations](pyspark-03-joins-and-aggregations.ipynb) | Join types, broadcast joins, `groupBy` and conditional aggregates |
| 4 | [Windows and nested data](pyspark-04-windows-and-nested-data.ipynb) | Window functions, dates and time zones, nested JSON and arrays |
| 5 | [Writing, pipelines and quality](pyspark-05-writing-pipelines-and-quality.ipynb) | Writing tables, incremental loads, testing, bad records, UDFs |
| 6 | [Structured streaming](pyspark-06-structured-streaming.ipynb) | Streams from files, watermarks, `foreachBatch`, `availableNow` |
| 7 | [Monitoring, performance and tuning](pyspark-07-monitoring-and-performance.ipynb) | Plans, the Spark UI, measuring a change, skew and salting, AQE, joins, file layout, caching, executor sizing, tuning exercises |
| 8 | [Real-world use cases and interviews](pyspark-08-real-world-and-interviews.ipynb) | Production patterns and interview answers |
