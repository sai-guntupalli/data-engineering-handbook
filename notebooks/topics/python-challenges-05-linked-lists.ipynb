````python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def make_list(values):
    """Build a linked list from a Python list and return its head (None when empty)."""
    head = None
    for value in reversed(values):
        head = ListNode(value, head)
    return head

def to_list(head):
    """Read a linked list back into a Python list."""
    values = []
    while head is not None:
        values.append(head.val)
        head = head.next
    return values
````

def reverse_list(head):
    # Approach: walk the list once, pointing each node backwards as you go.
    # Save the next node first, because reversing a pointer would otherwise lose the rest.
    # previous ends as the old tail, which is the new head.
    # Complexity: O(n) time, O(1) space.
    previous = None
    while head is not None:
        following = head.next             # keep the rest of the list before we break the link
        head.next = previous              # point this node backwards
        previous, head = head, following  # advance both pointers
    return previous
```

def merge_two_sorted(first, second):
    # Approach: a dummy head node avoids special-casing the first node.
    # Repeatedly take the smaller front node and attach it to the tail.
    # Once one list is exhausted, attach the rest of the other list as it is.
    # Complexity: O(m + n) time, O(1) extra space (nodes are reused, not copied).
    dummy = ListNode()
    tail = dummy
    while first is not None and second is not None:
        if first.val <= second.val:      # <= keeps the merge stable
            tail.next, first = first, first.next
        else:
            tail.next, second = second, second.next
        tail = tail.next
    tail.next = first if first is not None else second   # leftover nodes are already sorted
    return dummy.next

def has_cycle(head):
    # Approach: Floyd's tortoise and hare. The fast pointer gains one step on the slow
    # pointer each round, so inside a cycle it must eventually land on the slow pointer.
    # Without a cycle, the fast pointer reaches the end first.
    # Complexity: O(n) time, O(1) space.
    slow = fast = head
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:                  # compare identity: we want the same node
            return True
    return False
```

def remove_nth_from_end(head, n):
    # Approach: two pointers with a gap of n nodes. When the lead reaches the last node,
    # the trailing pointer sits just before the node to delete.
    # A dummy node in front handles removing the head itself.
    # Complexity: O(L) time for L nodes, O(1) space, one pass.
    dummy = ListNode(0, head)
    lead = trail = dummy
    for _ in range(n):                    # create the gap of n nodes
        lead = lead.next
    while lead.next is not None:          # move both until lead is the last node
        lead, trail = lead.next, trail.next
    trail.next = trail.next.next          # unlink the target node
    return dummy.next
```

def add_two_numbers(first, second):
    # Approach: add digit by digit, the same way you add on paper, from the least significant.
    # The lists are already in that order. Keep a carry for the next digit.
    # The loop also runs while carry is non-zero, which adds the final extra digit.
    # Complexity: O(max(m, n)) time, O(max(m, n)) space for the result.
    dummy = ListNode()
    tail = dummy
    carry = 0
    while first is not None or second is not None or carry:
        total = carry
        if first is not None:
            total += first.val
            first = first.next
        if second is not None:
            total += second.val
            second = second.next
        carry, digit = divmod(total, 10)  # divmod gives the new carry and this digit at once
        tail.next = ListNode(digit)
        tail = tail.next
    return dummy.next
```
