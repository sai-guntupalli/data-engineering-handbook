This notebook is the entry point to the SQL learning path. The path has fourteen chapters: chapters 1 to 12 each teach one topic, chapter 13 is a set of interview drills,
and chapter 14 covers interview concepts. Every example is runnable Spark SQL on small in-memory tables, and every result is
asserted, so a wrong answer fails loudly.

Work through the notebooks in order. Each one is self-contained, so you can also jump straight to a topic.

`UNPIVOT` was dropped: the same reshaping is covered by `PIVOT`, and it does not apply to the data-engineering interview questions.
`DELETE` and `UPDATE` were rewritten as "keep the survivors" and "apply the latest update with a fallback", because Spark
temporary views are read-only. The reference statements for table formats are in
[Views and changing data](sql-12-views-and-changing-data.ipynb).
