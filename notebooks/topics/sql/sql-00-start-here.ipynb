## Who this is for

- **New to SQL:** start at chapter 1 and go in order. Chapters 1 to 8 are the core.
- **Preparing for interviews:** skim chapters 2, 3, 8 and 9, then do the timed problems in chapter 13 and the
  questions in chapter 14.
- **Already fluent:** use the chapter table as an index. Every notebook repeats its own setup, so any chapter runs on
  its own.

You need basic Python only to read the setup cell: every query sits inside `spark.sql("""...""")`.

## Before you start

- **Run the cells in order.** The first code cell installs PySpark when it is missing, so the notebooks
  run in Colab without any setup. On your own machine PySpark 4.0 needs Java 17 or later.
- **Answers are checked.** Practice questions come with an answer cell that runs the query and asserts its result.
  Write your own query first, then run the answer cell.
- **The dialect is Spark SQL.** Most of what you learn carries over to PostgreSQL, Snowflake and BigQuery. The
  chapters point out Spark-only syntax where it comes up, such as `PIVOT`, the `<=>` null-safe equality operator, `explode` and
  0-based array indexing. Spark SQL has no `QUALIFY`, no indexes and no `UPDATE` on temporary views; chapters 8, 11
  and 12 show what to use instead.
- **For a quick lookup**, the [SQL reference](sql-reference.ipynb) collects syntax, dialect differences and
  engine internals in one place. It runs on DuckDB.
