## How the notebooks work

- **Spark SQL 4.0.** Every notebook runs on a local Spark session in Colab or locally (install the `notebooks` extra and Java 17 or 21). Notebooks 10 and 11 start a Delta Lake session for `UPDATE`, `MERGE`, constraints and time travel; the Delta jars download on first use.
- **Shared sample data.** All notebooks define the same `customers`, `orders`, `order_items`, `products`, `events` and `employees` temporary views (matching the PySpark cookbook), so you only learn the data once.
- **Practice format.** Each exercise names the reference entries it practises, shows the exact expected output and hides a solution you can open when you are done or stuck. CI checks every solution against its published output, re-runs it against edge-case tables, and checks query-plan properties in notebook 12.
