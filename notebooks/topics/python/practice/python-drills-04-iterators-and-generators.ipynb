```python
def running_total(numbers):
    total = 0
    for number in numbers:
        total += number
        yield total
```

<details>
<summary><b>Show solution</b></summary>

```python
def chunks(items, size):
    batch = []
    for item in items:
        batch.append(item)
        if len(batch) == size:
            yield batch
            batch = []
    if batch:
        yield batch
```

```python
def first_evens(stream, n):
    found = []
    if n <= 0:
        return found
    for value in stream:
        if value % 2 == 0:
            found.append(value)
            if len(found) == n:
                break
    return found
```

<details>
<summary><b>Show solution</b></summary>

```python
def flatten(nested):
    for item in nested:
        if isinstance(item, list):
            yield from flatten(item)
        else:
            yield item
```

<summary><b>Show solution</b></summary>

```python
def pairwise(values):
    iterator = iter(values)
    try:
        previous = next(iterator)
    except StopIteration:
        return
    for current in iterator:
        yield previous, current
        previous = current
```

```python
from itertools import groupby

def sum_by_key(pairs):
    return [(key, sum(value for _, value in group)) for key, group in groupby(pairs, key=lambda pair: pair[0])]
```
