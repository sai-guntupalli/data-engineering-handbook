def is_palindrome(text):
    # Approach: two pointers move inward from both ends.
    # Skip characters that are not letters or digits, compare the rest case-insensitively.
    # Complexity: O(n) time, O(1) space.
    left, right = 0, len(text) - 1
    while left < right:
        if not text[left].isalnum():        # skip punctuation and spaces on the left
            left += 1
        elif not text[right].isalnum():     # ...and on the right
            right -= 1
        elif text[left].lower() != text[right].lower():
            return False
        else:
            left += 1
            right -= 1
    return True
    # Tip: the whole string is never copied; cleaning it first would cost O(n) extra space.
```

def three_sum(nums):
    # Approach: sort, then for each first value run a two-pointer search for the other two.
    # Sorting puts the candidates in order, so a too-small sum means move left up,
    # and a too-large sum means move right down.
    # Duplicates are skipped at every level so each triplet appears once.
    # Complexity: O(n^2) time, O(1) extra space (ignoring the output and the sort).
    ordered = sorted(nums)
    triplets = []
    for i in range(len(ordered) - 2):
        if i and ordered[i] == ordered[i - 1]:      # same first value as before: already done
            continue
        left, right = i + 1, len(ordered) - 1
        while left < right:
            total = ordered[i] + ordered[left] + ordered[right]
            if total < 0:
                left += 1
            elif total > 0:
                right -= 1
            else:
                triplets.append([ordered[i], ordered[left], ordered[right]])
                left += 1
                while left < right and ordered[left] == ordered[left - 1]:
                    left += 1               # skip duplicate second values
    return triplets
```

def max_area(heights):
    # Approach: start with the widest container, then move the shorter wall inward.
    # The area is limited by the shorter wall, so keeping the taller wall and moving the
    # shorter one is the only move that could still find a larger area.
    # Complexity: O(n) time, O(1) space.
    left, right = 0, len(heights) - 1
    best = 0
    while left < right:
        width = right - left
        best = max(best, width * min(heights[left], heights[right]))
        if heights[left] < heights[right]:  # the left wall limits the area: replace it
            left += 1
        else:
            right -= 1
    return best

def length_of_longest_substring(text):
    # Approach: a sliding window [left, right] that never contains a repeat.
    # last_seen remembers where each character last appeared. When a repeat falls inside
    # the window, jump left just past its previous position.
    # Complexity: O(n) time, O(min(n, alphabet)) space.
    last_seen = {}
    left = 0
    best = 0
    for right, char in enumerate(text):
        # Only a repeat at or after left matters; an older one is already outside the window.
        if char in last_seen and last_seen[char] >= left:
            left = last_seen[char] + 1
        last_seen[char] = right
        best = max(best, right - left + 1)
    return best
```

from collections import Counter

def min_window(s, t):
    # Approach: a sliding window that grows on the right until it covers t, then shrinks
    # on the left as far as it can while still covering t.
    # need[c] = how many more of c the window needs (goes negative when the window has extras).
    # missing = total characters still needed, so the window is valid exactly when missing == 0.
    # Complexity: O(|s| + |t|) time, O(alphabet) space.
    need = Counter(t)
    missing = len(t)
    best_length, best = float("inf"), ""
    left = 0
    for right, char in enumerate(s, 1):     # window is s[left:right]
        if need[char] > 0:                  # this character was still needed
            missing -= 1
        need[char] -= 1
        while missing == 0:                 # window covers t: record it, then try to shrink
            if right - left < best_length:
                best_length, best = right - left, s[left:right]
            need[s[left]] += 1
            if need[s[left]] > 0:           # removing it made the window invalid
                missing += 1
            left += 1
    return best
```
