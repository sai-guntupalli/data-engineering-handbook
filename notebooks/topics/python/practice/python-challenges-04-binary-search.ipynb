def search(nums, target):
    # Approach: each comparison with the middle value throws away half the remaining range.
    # Complexity: O(log n) time, O(1) space.
    low, high = 0, len(nums) - 1
    while low <= high:                    # the range [low, high] is still a candidate
        mid = (low + high) // 2
        if nums[mid] == target:
            return mid
        if nums[mid] < target:
            low = mid + 1                 # target is to the right of mid
        else:
            high = mid - 1                # target is to the left of mid
    return -1
```

def search_rotated(nums, target):
    # Approach: binary search where at least one half is always sorted.
    # If the left half is sorted and the target lies inside its range, search there;
    # otherwise the target can only be in the other half. The right half is handled the same way.
    # Complexity: O(log n) time, O(1) space.
    low, high = 0, len(nums) - 1
    while low <= high:
        mid = (low + high) // 2
        if nums[mid] == target:
            return mid
        if nums[low] <= nums[mid]:                    # left half [low..mid] is sorted
            if nums[low] <= target < nums[mid]:       # target inside that sorted range
                high = mid - 1
            else:
                low = mid + 1
        else:                                         # right half [mid..high] is sorted
            if nums[mid] < target <= nums[high]:      # target inside that sorted range
                low = mid + 1
            else:
                high = mid - 1
    return -1
```

def find_min_rotated(nums):
    # Compare nums[mid] with nums[high]. If it is bigger, the minimum lies
    # to the right of mid; otherwise keep mid in the remaining search range.
    # if mid is bigger than nums[high], the rotation point (and the minimum) is to the right.
    # Complexity: O(log n) time, O(1) space.
    low, high = 0, len(nums) - 1
    while low < high:                     # stop when one candidate is left
        mid = (low + high) // 2
        if nums[mid] > nums[high]:
            low = mid + 1                 # the drop is to the right of mid
        else:
            high = mid                    # mid could itself be the minimum: keep it
    return nums[low]

def min_eating_speed(piles, h):
    # Approach: binary search on the answer. If speed s finishes in time, any faster speed
    # does too, so the valid speeds form a range whose lower edge we search for.
    # Hours for one pile at speed s = ceil(pile / s), written as integer arithmetic.
    # Complexity: O(n * log(max pile)) time, O(1) space.
    low, high = 1, max(piles)             # the slowest and the fastest sensible speeds
    while low < high:
        speed = (low + high) // 2
        hours = sum((pile + speed - 1) // speed for pile in piles)   # ceil division
        if hours <= h:
            high = speed                  # feasible: try slower
        else:
            low = speed + 1               # too slow: must go faster
    return low
```

from bisect import bisect_left, bisect_right

def search_range(nums, target):
    # Approach: two binary searches. bisect_left finds the first index where target could go,
    # bisect_right finds the first index after the last target. Their difference is the count.
    # Complexity: O(log n) time, O(1) space.
    first = bisect_left(nums, target)
    last = bisect_right(nums, target) - 1     # last index holding target
    if first <= last:                          # the target actually occurs
        return [first, last]
    return [-1, -1]
```
