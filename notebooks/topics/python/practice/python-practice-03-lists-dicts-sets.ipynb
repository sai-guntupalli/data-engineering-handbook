**Sample calls**

| call | expected |
| --- | --- |
| `duplicates([1, 2, 1, 3, 2, 1])` | `[1, 2]` |
| `duplicates([1, 2, 3])` | `[]` |
| `duplicates([])` | `[]` |

```python
def duplicates(items):
    # your code here
    pass
```

```python
def duplicates(items):
    seen, dupes = set(), {}
    for item in items:
        if item in seen:
            dupes.setdefault(item, None)
        seen.add(item)
    return list(dupes)
```

```python
def two_sum(nums, target):
    seen = {}
    for j, value in enumerate(nums):
        if target - value in seen:
            return [seen[target - value], j]
        seen.setdefault(value, j)
    return None
```

| `get_path({"a": {}}, "a.b", "x")` | `'x'` |
| `get_path({"items": [{"sku": "A"}]}, "items.0.sku")` | `'A'` |

```python
def get_path(data, path, default=None):
    current = data
    for part in path.split("."):
        if isinstance(current, dict) and part in current:
            current = current[part]
        elif isinstance(current, list) and part.lstrip("-").isdigit() and -len(current) <= int(part) < len(current):
            current = current[int(part)]
        else:
            return default
    return current
```
