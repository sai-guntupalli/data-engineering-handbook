```python
def fizzbuzz(n):
    out = []
    for i in range(1, n + 1):
        if i % 15 == 0:
            out.append("FizzBuzz")
        elif i % 3 == 0:
            out.append("Fizz")
        elif i % 5 == 0:
            out.append("Buzz")
        else:
            out.append(str(i))
    return out
```

```python
def first_gap(ids):
    expected = 1
    for value in ids:
        if value == expected:
            expected += 1
        else:
            break
    return expected
```

```python
def format_row(values, *, sep=",", quote=False):
    items = [str(v) for v in values]
    if quote:
        items = [f'"{item}"' for item in items]
    return sep.join(items)
```

<details>

```python
def pipeline(*steps):
    def run(value):
        for step in steps:
            value = step(value)
        return value
    return run
```

```python
import functools

def count_calls(fn):
    @functools.wraps(fn)
    def wrapper(*args, **kwargs):
        wrapper.calls += 1
        return fn(*args, **kwargs)
    wrapper.calls = 0
    return wrapper
```
