```python
import re
import unicodedata

def slugify(text):
    ascii_text = "".join(c for c in unicodedata.normalize("NFKD", text) if not unicodedata.combining(c))
    return re.sub(r"[^a-z0-9]+", "-", ascii_text.lower()).strip("-")
```

```python
def parse_pairs(text):
    result = {}
    for item in text.split(";"):
        if not item.strip():
            continue
        key, _, value = item.partition("=")
        result[key.strip()] = value.strip()
    return result
```

```python
import re

HEAD = re.compile(r"(?P<date>\d{4}-\d{2}-\d{2}) [\d:]{8} (?P<level>[A-Z]+)\b")

def parse_log(line):
    m = HEAD.match(line)
    if not m:
        return None
    record = {"date": m.group("date"), "level": m.group("level")}
    record.update(re.findall(r"(\w+)=(\S+)", line[m.end():]))
    return record
```

```python
import re

def hashtags(text):
    seen = {}
    for tag in re.findall(r"#(\w+)", text):
        seen.setdefault(tag.casefold(), None)
    return list(seen)
```

```python
import unicodedata

def name_key(name):
    decomposed = unicodedata.normalize("NFKD", name)
    stripped = "".join(c for c in decomposed if not unicodedata.combining(c))
    return " ".join(stripped.casefold().split())
```
