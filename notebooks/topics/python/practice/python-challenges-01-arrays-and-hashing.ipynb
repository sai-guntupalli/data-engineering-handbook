def two_sum(nums, target):
    # Approach: one pass with a hash map from value -> index of values already seen.
    # For each number, the partner we need is target - value. If it was seen earlier,
    # we have the pair; otherwise store this value for later numbers to find.
    # Complexity: O(n) time, O(n) space.
    seen = {}
    for index, value in enumerate(nums):
        if target - value in seen:
            return [seen[target - value], index]
        seen[value] = index
    # Tip: checking before storing means a number cannot pair with itself.
```

from collections import defaultdict

def group_anagrams(words):
    # Approach: two words are anagrams exactly when their sorted letters match.
    # Sorted letters make a shared key; the dict collects every word under its key.
    # Complexity: O(n * k log k) time for n words of length k, O(n * k) space.
    groups = defaultdict(list)
    for word in words:
        key = "".join(sorted(word))      # e.g. "eat" and "tea" both give "aet"
        groups[key].append(word)
    return list(groups.values())
    # Tip: a letter-count tuple works too and avoids the sort; it is O(n * k) overall.
```

from collections import Counter

def top_k_frequent(nums, k):
    # Approach: count every value, then take the k values with the highest counts.
    # Counter.most_common(k) does the selection for us (it uses a heap internally).
    # Complexity: O(n) to count, O(n log k) to select the top k.
    return [value for value, _ in Counter(nums).most_common(k)]
    # Tip: a bucket array indexed by count gives O(n) overall when you need that bound.

def product_except_self(nums):
    # Approach: answer[i] = (product of everything left of i) * (product of everything right of i).
    # Pass 1 (left to right) fills answer with the left products.
    # Pass 2 (right to left) multiplies in the right products using one running value.
    # No division is needed, so zeros are handled naturally.
    # Complexity: O(n) time, O(1) extra space (the output list does not count).
    result = [1] * len(nums)
    prefix = 1                              # product of nums[0 .. i-1]
    for i, value in enumerate(nums):
        result[i] = prefix
        prefix *= value
    suffix = 1                              # product of nums[i+1 .. end]
    for i in range(len(nums) - 1, -1, -1):
        result[i] *= suffix
        suffix *= nums[i]
    return result
```

def longest_consecutive(nums):
    # Approach: put every value in a set so membership is O(1).
    # Only start counting from a value whose predecessor (value - 1) is absent.
    # That start is the beginning of its run, so each number is visited at most twice.
    # Complexity: O(n) time, O(n) space.
    values = set(nums)
    best = 0
    for value in values:
        if value - 1 not in values:         # not the middle of a run: skip it
            length = 1
            while value + length in values:
                length += 1
            best = max(best, length)
    return best
    # Tip: without the "start of run" check, the loop would be O(n^2) on long runs.
```
