````python
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def make_tree(level_values):
    """Build a tree from a level-order list, where None marks a missing child."""
    if not level_values or level_values[0] is None:
        return None
    root = TreeNode(level_values[0])
    queue = deque([root])
    index = 1
    while queue and index < len(level_values):
        node = queue.popleft()
        if index < len(level_values) and level_values[index] is not None:
            node.left = TreeNode(level_values[index])
            queue.append(node.left)
        index += 1
        if index < len(level_values) and level_values[index] is not None:
            node.right = TreeNode(level_values[index])
            queue.append(node.right)
        index += 1
    return root

def level_order(root):
    """Read a tree back as a level-order list, with None for missing children and no trailing Nones."""
    values, queue = [], deque([root])
    while queue:
        node = queue.popleft()
        if node is None:
            values.append(None)
        else:
            values.append(node.val)
            queue.append(node.left)
            queue.append(node.right)
    while values and values[-1] is None:
        values.pop()
    return values
````

    # Approach: a tree's depth is 1 plus the deeper of its two subtrees.
    # An empty tree is depth 0, which is also the base case of the recursion.

def invert_tree(root):
    # Approach: swap the two children of every node. Invert both subtrees first, then
    # swap them. Because each node is visited once, the order does not change the result.
    # Complexity: O(n) time, O(h) space for the recursion stack.

def is_valid_bst(root, low=float("-inf"), high=float("inf")):
    # Approach: every node must sit strictly between the bounds set by its ancestors.
    # Going left tightens the upper bound to the parent's value; going right tightens the lower.
    # Checking only parent-child pairs is not enough: a grandchild can break an ancestor's bound.
    # Complexity: O(n) time, O(h) space.

        return False

def lowest_common_ancestor(root, p, q):
    # Approach: BST ordering tells us which way to go. If both targets are smaller than the
    # current node, the answer is on the left; if both are larger, on the right.
    # Otherwise the targets split here (or one equals this node), so this node is the answer.
    # Complexity: O(h) time, O(1) space for the iterative version.
    node = root
    while node is not None:
        if p.val < node.val and q.val < node.val:
            node = node.left
        elif p.val > node.val and q.val > node.val:
            node = node.right
        else:
            return node                   # the paths to p and q split here
    return None
```

from collections import deque

def level_order_groups(root):
    # Approach: breadth-first search with a queue. Before processing a level, record how many
    # nodes it holds (len(queue)); processing exactly that many yields one level as a group.
    # Complexity: O(n) time, O(w) space, where w is the widest level.
    groups = []
    queue = deque([root] if root else [])
    while queue:
        level = []
        for _ in range(len(queue)):       # exactly this level's nodes, not the children
            node = queue.popleft()
            level.append(node.val)
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        groups.append(level)
    return groups
```
