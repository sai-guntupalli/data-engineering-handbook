## How the notebooks work

- **Run anywhere.** Every notebook opens in Colab and runs top to bottom with the standard library only (Python 3.12 or later; written against 3.13).
- **Practice format.** Each exercise names the reference entries it practises, shows sample calls with their exact results, and hides a solution you can open when you are done or stuck. CI runs every solution against hidden edge-case checks and confirms that a wrong answer fails them.
